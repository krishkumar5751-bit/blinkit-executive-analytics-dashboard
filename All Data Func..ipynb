# Table-3  "Orders"

```python
orders=pd.read_csv("orders.csv")
print(orders)
```

```python
# Check Null
orders.isnull().sum()
```

```python
#  Datetime conversion & Date/Time features

orders['order_timestamp'] = pd.to_datetime(orders['order_timestamp'])
orders['order_date'] = orders['order_timestamp'].dt.date
orders['order_hour'] = orders['order_timestamp'].dt.hour
orders['day_of_week'] = orders['order_timestamp'].dt.day_name()
orders['is_weekend'] = orders['order_timestamp'].dt.dayofweek.isin([5, 6]).astype(int)
```

```python
# Rider ID nulls ko explicit category dena

orders['rider_id'] = orders['rider_id'].fillna('Unassigned')
```

```python
# Again Check Null
orders.isnull().sum()
```

```python
orders.to_csv('cleaned_orders.csv', index=False)
print("Orders table successfully cleaned & saved as cleaned_orders.csv!")
```
